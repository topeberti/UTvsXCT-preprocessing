```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

from pathlib import Path
from preprocess_tools import io, sample_isolater
```

## Configuration

| Variable | Description |
|---|---|
| `VOLUME_PATH` | Single multi-page TIFF **or** folder of per-slice TIFFs |
| `SAVING_FOLDER` | Output folder — one sub-folder per sample will be created inside it |
| `N_SAMPLES` | Number of physical samples in the scan |
| `SAMPLE_NAMES` | Optional list of names for the output folders. If `None`, samples are named `sample_00`, `sample_01`, … |

```python
VOLUME_PATH   = Path('/home/jorgecabrejas/Dev/Data/equalized/20_1+2+4-21+1+2+3_25um_eq_8b')
SAVING_FOLDER = Path('/home/jorgecabrejas/Desktop/isolated_samples')
N_SAMPLES     = 6
SAMPLE_NAMES  = [f"20_{i}" for i in (1, 2, 4)] + [f"21_{i}" for i in range(1, 4)]
```

## Load

```python
volume = io.load_tif(VOLUME_PATH)
print(f'Loaded  shape={volume.shape}  dtype={volume.dtype}  range=[{volume.min()}, {volume.max()}]')
```

## Isolate samples

```python
isolated_volumes = sample_isolater.isolate_samples(volume, N_SAMPLES)

for i, vol in enumerate(isolated_volumes):
    print(f'Sample {i}: shape={vol.shape}')
```

## Save

```python
names = SAMPLE_NAMES if SAMPLE_NAMES is not None else [f'sample_{i:02d}' for i in range(N_SAMPLES)]

for name, vol in zip(names, isolated_volumes):
    out_path = SAVING_FOLDER / name
    out_path.mkdir(parents=True, exist_ok=True)
    print(f'Saving {name} -> {out_path}')
    io.save_tif(str(out_path), vol)
```
